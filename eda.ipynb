## gh-aw corpus (`data/gh_aw_workflows`)

Load `ai_agentic_workflows.csv` and attach each workflow's local `.md` / `.lock.yml` text (body after the ingest header) for analysis.

```python
from pathlib import Path

GH_AW_ROOT = Path("data/gh_aw_workflows")
MD_DIR = GH_AW_ROOT / "workflow-markdowns"
YML_DIR = GH_AW_ROOT / "workflow-yamls"
HEADER_SEP = "=" * 80


def _safe_stem(repo: str, workflow: str) -> str:
    return f"{repo.replace('/', '_')}_{workflow.replace('/', '_')}"


def _read_collected(path: Path) -> str | None:
    """Read a collected file; return body after the ingest header, or full text."""
    if not path.is_file():
        return None
    text = path.read_text(encoding="utf-8", errors="replace")
    if HEADER_SEP in text:
        return text.split(HEADER_SEP, 1)[1].lstrip("\n")
    return text


df_gh_aw = pd.read_csv(GH_AW_ROOT / "ai_agentic_workflows.csv")
stems = [_safe_stem(r, w) for r, w in zip(df_gh_aw["repo"], df_gh_aw["workflow_name"])]
df_gh_aw["md_file"] = [str(MD_DIR / f"{s}.md") for s in stems]
df_gh_aw["lock_file"] = [str(YML_DIR / f"{s}.lock.yml") for s in stems]
df_gh_aw["md_content"] = [ _read_collected(MD_DIR / f"{s}.md") for s in stems ]
df_gh_aw["lock_content"] = [ _read_collected(YML_DIR / f"{s}.lock.yml") for s in stems ]

missing_md = df_gh_aw["md_content"].isna().sum()
missing_lock = df_gh_aw["lock_content"].isna().sum()
(
    df_gh_aw.shape,
    df_gh_aw["repo"].nunique(),
    f"missing md={missing_md}",
    f"missing lock={missing_lock}",
)
```

# Analyze github agentic workflow

## Task-category classification

Simple flat labels: `issue triage`, `pr review`, `documentation`, `ci repair`, ...


```python
# Zero-shot task categories for gh-aw workflows
import re
from pathlib import Path

import torch
from transformers import pipeline
from tqdm.auto import tqdm

# Simple flat labels (stored in dataframe)
TASK_CATEGORIES = [
    "issue triage",
    "pr review",
    "documentation",
    "ci repair",
    "security",
    "release",
    "testing",
    "bug fix",
    "content",
    "planning",
    "monitoring",
    "other",
]

# Verbalized phrases for zero-shot NLI (short labels like "testing" confuse the model)
ZS_VERBAL = {
    "issue triage": "triaging and labeling GitHub issues",
    "pr review": "reviewing pull requests and source code",
    "documentation": "writing or updating documentation",
    "ci repair": "investigating and fixing CI or build failures",
    "security": "security, dependency, or supply chain review",
    "release": "releases, versioning, or changelogs",
    "testing": "improving test coverage or test suite quality",
    "bug fix": "implementing bug fixes in code",
    "content": "generating reports, digests, or content summaries",
    "planning": "research, planning, or specification work",
    "monitoring": "monitoring, observability, or operational alerts",
    "other": "something else",
}
ZS_LABELS = list(ZS_VERBAL.values())
ZS_TO_SIMPLE = {v: k for k, v in ZS_VERBAL.items()}

HYPOTHESIS = "This GitHub agentic workflow is primarily about {}."
ZS_MODEL = "MoritzLaurer/deberta-v3-base-zeroshot-v2.0"
MAX_CHARS = 1800


def _frontmatter_description(md: str) -> str:
    m = re.match(r"^---\s*\n(.*?)\n---\s*\n", md, flags=re.S)
    if not m:
        return ""
    block = m.group(1)
    for key in ("description", "name"):
        km = re.search(rf"(?m)^{key}:\s*[\"\']?(.*?)[\"\']?\s*$", block)
        if km:
            return km.group(1).strip()
    return ""


def classify_text(row) -> str:
    name = str(row.get("workflow_name") or "")
    md = str(row.get("md_content") or "")
    desc = _frontmatter_description(md)
    body = re.sub(r"^---\s*\n.*?\n---\s*\n", "", md, count=1, flags=re.S).strip()
    parts = [f"Workflow: {name}"]
    if desc:
        parts.append(f"Description: {desc}")
    if body:
        parts.append(body[:MAX_CHARS])
    return "\n\n".join(parts)


df_task = df_gh_aw.copy()
print(f"classifying df_gh_aw: {len(df_task)} workflows")
print("categories:", TASK_CATEGORIES)

device = 0 if torch.cuda.is_available() else ("mps" if getattr(torch.backends, "mps", None) and torch.backends.mps.is_available() else -1)
zs = pipeline(
    "zero-shot-classification",
    model=ZS_MODEL,
    device=device,
)
print(f"model={ZS_MODEL} device={device}")
```

```python
texts = [classify_text(r) for _, r in df_task.iterrows()]

labels, scores, top3 = [], [], []
for text in tqdm(texts, desc="zero-shot"):
    out = zs(
        text,
        candidate_labels=ZS_LABELS,
        hypothesis_template=HYPOTHESIS,
        multi_label=False,
    )
    labels.append(ZS_TO_SIMPLE[out["labels"][0]])
    scores.append(float(out["scores"][0]))
    top3.append(
        {
            ZS_TO_SIMPLE[lab]: float(sc)
            for lab, sc in zip(out["labels"][:3], out["scores"][:3])
        }
    )

df_task["task_category"] = labels
df_task["task_category_score"] = scores
df_task["task_category_top3"] = top3

df_gh_aw["task_category"] = labels
df_gh_aw["task_category_score"] = scores
df_gh_aw["task_category_top3"] = top3

out_csv = Path("data/gh_aw_workflows/task_categories_zeroshot.csv")
df_task[
    ["repo", "workflow_name", "task_category", "task_category_score"]
].to_csv(out_csv, index=False)
print(f"wrote {out_csv}")
df_task["task_category"].value_counts(dropna=False)
```

## LLM classification (DeepSeek)

LiteLLM JSON mode + `deepseek-v4-flash`. Parallel requests via thread pool (`LLM_CONCURRENCY`, default 16).

Set `DEEPSEEK_API_KEY` in `.env`. Optional: `LLM_CONCURRENCY=32`.


```python
# LLM task-category classifier (DeepSeek via LiteLLM)
import json
import os
import time
from pathlib import Path

import pandas as pd
from dotenv import load_dotenv
from litellm import completion
from pydantic import BaseModel, Field, field_validator
from tqdm.auto import tqdm

load_dotenv("gh_data_ingestion/.env")
load_dotenv(".env")

if "TASK_CATEGORIES" not in dir():
    TASK_CATEGORIES = [
        "issue triage",
        "pr review",
        "documentation",
        "ci repair",
        "security",
        "release",
        "testing",
        "bug fix",
        "content",
        "planning",
        "monitoring",
        "other",
    ]

DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL", "deepseek-v4-flash")
LLM_MAX_CHARS = 6000
LLM_PROGRESS = Path("data/gh_aw_workflows/.progress/llm_task_categories.jsonl")
LLM_OUT_CSV = Path("data/gh_aw_workflows/task_categories_llm.csv")
LLM_CONCURRENCY = int(os.getenv("LLM_CONCURRENCY", "32"))

SYSTEM_PROMPT = f"""You classify GitHub Agentic Workflows (gh-aw) by primary task.

Return exactly ONE category from this list:
{", ".join(TASK_CATEGORIES)}

Rules:
- issue triage: label/route GitHub issues
- pr review: review pull requests or code
- documentation: write/update docs
- ci repair: diagnose/fix failed CI or builds
- security: security or dependency review
- release: releases or changelogs
- testing: improve/audit tests only (not just because CI runs tests)
- bug fix: implement code fixes
- content: digests/reports/summaries
- planning: specs/research/planning
- monitoring: ops monitoring/alerts
- other: none of the above

Respond with JSON only: {{"category": "...", "confidence": 0.0, "rationale": "..."}}
"""


class WorkflowTaskLabel(BaseModel):
    category: str
    confidence: float = Field(ge=0.0, le=1.0, description="0-1 confidence")
    rationale: str = Field(max_length=300)

    @field_validator("category")
    @classmethod
    def _valid_category(cls, v: str) -> str:
        v = v.strip().lower()
        if v not in TASK_CATEGORIES:
            raise ValueError(f"category must be one of {TASK_CATEGORIES}")
        return v


def llm_input_text(row) -> str:
    if "classify_text" in dir():
        return classify_text(row)
    import re

    name = str(row.get("workflow_name") or "")
    md = str(row.get("md_content") or "")
    m = re.match(r"^---\s*\n(.*?)\n---\s*\n", md, flags=re.S)
    desc = ""
    if m:
        for key in ("description", "name"):
            km = re.search(rf"(?m)^{key}:\s*[\"\']?(.*?)[\"\']?\s*$", m.group(1))
            if km:
                desc = km.group(1).strip()
                break
    body = re.sub(r"^---\s*\n.*?\n---\s*\n", "", md, count=1, flags=re.S).strip()
    parts = [f"Workflow: {name}"]
    if desc:
        parts.append(f"Description: {desc}")
    if body:
        parts.append(body[:LLM_MAX_CHARS])
    return "\n\n".join(parts)


def llm_classify_one(text: str) -> WorkflowTaskLabel:
    """JSON mode + thinking disabled (v4-flash rejects tool_choice in thinking mode)."""
    if not os.getenv("DEEPSEEK_API_KEY"):
        raise RuntimeError("Set DEEPSEEK_API_KEY in gh_data_ingestion/.env")
    resp = completion(
        model=DEEPSEEK_MODEL,
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": text},
        ],
        response_format={"type": "json_object"},
        reasoning_effort="none",
        temperature=0,
        max_tokens=512,
        extra_body={"thinking": {"type": "disabled"}},
    )
    raw = resp.choices[0].message.content
    return WorkflowTaskLabel.model_validate_json(raw)


def _pair_key(repo: str, workflow: str) -> str:
    return f"{repo}::{workflow}"


def load_llm_done(path: Path) -> dict[str, dict]:
    """Load successful labels only; failed rows will be retried."""
    done: dict[str, dict] = {}
    if not path.exists():
        return done
    with path.open(encoding="utf-8") as fh:
        for line in fh:
            line = line.strip()
            if not line:
                continue
            rec = json.loads(line)
            if not (rec.get("repo") and rec.get("workflow_name")):
                continue
            if rec.get("error") or rec.get("task_category") is None:
                continue
            done[_pair_key(rec["repo"], rec["workflow_name"])] = rec
    return done


print(f"model={DEEPSEEK_MODEL}")
print(f"categories={TASK_CATEGORIES}")
print(f"resume file={LLM_PROGRESS}")
print(f"concurrency={LLM_CONCURRENCY}")

def classify_row(row) -> dict:
    """Classify one workflow; safe to call from thread pool."""
    text = llm_input_text(row)
    try:
        label = llm_classify_one(text)
        return {
            "repo": row["repo"],
            "workflow_name": row["workflow_name"],
            "task_category": label.category,
            "task_category_score": label.confidence,
            "task_category_rationale": label.rationale,
        }
    except Exception as exc:
        return {
            "repo": row["repo"],
            "workflow_name": row["workflow_name"],
            "task_category": None,
            "task_category_score": None,
            "task_category_rationale": None,
            "error": str(exc),
        }
```

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import threading

LLM_PROGRESS.parent.mkdir(parents=True, exist_ok=True)
done = load_llm_done(LLM_PROGRESS)

pending_rows = [
    row
    for _, row in df_gh_aw.iterrows()
    if _pair_key(row["repo"], row["workflow_name"]) not in done
]
print(f"already labeled: {len(done)} / {len(df_gh_aw)} | pending: {len(pending_rows)} | workers={LLM_CONCURRENCY}")

new_labels = 0
errors = 0
write_lock = threading.Lock()

with LLM_PROGRESS.open("a", encoding="utf-8") as prog, ThreadPoolExecutor(max_workers=LLM_CONCURRENCY) as pool:
    futures = [pool.submit(classify_row, row) for row in pending_rows]
    for fut in tqdm(as_completed(futures), total=len(futures), desc="llm classify"):
        rec = fut.result()
        pk = _pair_key(rec["repo"], rec["workflow_name"])
        with write_lock:
            prog.write(json.dumps(rec, ensure_ascii=False) + "\n")
            prog.flush()
            if rec.get("task_category"):
                done[pk] = rec
                new_labels += 1
            else:
                errors += 1

records = list(load_llm_done(LLM_PROGRESS).values())
df_llm = pd.DataFrame(records)
df_llm[["repo", "workflow_name", "task_category", "task_category_score"]].to_csv(
    LLM_OUT_CSV, index=False
)

drop_cols = [c for c in ["task_category_llm", "task_category_llm_score", "task_category_llm_rationale"] if c in df_gh_aw.columns]
df_gh_aw = df_gh_aw.drop(columns=drop_cols)
df_gh_aw = df_gh_aw.merge(
    df_llm[["repo", "workflow_name", "task_category", "task_category_score", "task_category_rationale"]].rename(
        columns={
            "task_category": "task_category_llm",
            "task_category_score": "task_category_llm_score",
            "task_category_rationale": "task_category_llm_rationale",
        }
    ),
    on=["repo", "workflow_name"],
    how="left",
)

print(f"new this run={new_labels}, errors={errors}, wrote {LLM_OUT_CSV}")
df_gh_aw["task_category_llm"].value_counts(dropna=False)
```

## Uses skills? (boolean) — `df_iss_triage` only

`uses_skills=True` only for **path skills**, not inline `## skill:`:

1. **Frontmatter** — non-empty `skills:` list (local paths or `owner/repo/...@sha`)
2. **Body** — instructs the agent to use a skill **file/folder** (e.g. `.agents/skills/.../SKILL.md`)

Runs only on **issue triage** rows (`df_iss_triage`). Deterministic regex first; DeepSeek LLM for ambiguous cases.


```python
# Detect path-skill usage (frontmatter + body path; NOT inline ## skill:)
import json
import os
import re
import threading
import time
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path

import pandas as pd
from dotenv import load_dotenv
from litellm import completion
from pydantic import BaseModel, Field
from tqdm.auto import tqdm

load_dotenv("gh_data_ingestion/.env")
load_dotenv(".env")

SKILLS_PROGRESS = Path("data/gh_aw_workflows/.progress/skills_usage_issue_triage.jsonl")
SKILLS_OUT_CSV = Path("data/gh_aw_workflows/skills_usage_issue_triage.csv")
SKILLS_CONCURRENCY = int(os.getenv("LLM_CONCURRENCY", "16"))
DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL", "deepseek-v4-flash")

FRONTMATTER_RE = re.compile(r"^---\s*\n(.*?)\n---\s*\n", re.S)
# Path skill folders / SKILL.md (local); exclude bare "skill" words
SKILL_PATH_RE = re.compile(
    r"(?:"
    r"\.github/skills/"
    r"|\.agents/skills/"
    r"|\.claude/skills/"
    r"|\.apm/skills/"
    r"|/(?:skills)/[\w./-]+/SKILL\.md"
    r"|\bSKILL\.md\b"
    r"|skills:\s*\n\s*-\s+\S+"  # already covered by frontmatter, but ok in body
    r")",
    re.I,
)
# Instruction-like body cues near skill paths
BODY_MUST_USE_RE = re.compile(
    r"(?is)(?:read\s+and\s+follow|follow(?:\s+the)?\s+instructions?\s+in|"
    r"use\s+(?:the\s+)?|load\s+|apply\s+(?:the\s+)?|"
    r"must\s+use|according\s+to).{0,120}?"
    r"(?:\.github/skills/|\.agents/skills/|\.claude/skills/|\.apm/skills/|SKILL\.md)",
)
INLINE_SKILL_RE = re.compile(r"(?m)^#{2,3}\s*skill\s*:", re.I)
SKILL_WORD_RE = re.compile(r"\bskills?\b", re.I)


def _split_md(md: str) -> tuple[str, str]:
    md = str(md or "")
    m = FRONTMATTER_RE.match(md)
    if not m:
        return "", md
    return m.group(1), md[m.end() :]


def has_frontmatter_skills(md: str) -> bool:
    """True if YAML frontmatter has a non-empty skills: list."""
    fm, _ = _split_md(md)
    if not fm:
        return False
    # skills: [path] or skills:\n  - path
    if re.search(r"(?m)^skills\s*:\s*\[\s*\]\s*$", fm):
        return False
    if re.search(r"(?m)^skills\s*:\s*\[\s*[^\]]+\S", fm):
        return True
    m = re.search(r"(?m)^skills\s*:\s*(.*)$", fm)
    if not m:
        return False
    rest = m.group(1).strip()
    if rest and rest not in ("[]", "~", "null", "None"):
        # inline scalar / flow list
        if not rest.startswith("#"):
            return True
    # block list under skills:
    lines = fm.splitlines()
    for i, line in enumerate(lines):
        if re.match(r"^skills\s*:\s*$", line) or re.match(r"^skills\s*:\s*#", line):
            for nxt in lines[i + 1 :]:
                if re.match(r"^\S", nxt):  # next top-level key
                    break
                if re.match(r"^\s+-\s+\S+", nxt):
                    return True
            break
    return False


def has_body_skill_path(md: str) -> bool:
    """True if body clearly points at a skill file/folder (not ## skill: inline)."""
    _, body = _split_md(md)
    if not body:
        return False
    # Strip inline skill blocks so they don't count
    body_wo_inline = INLINE_SKILL_RE.split(body)[0] if INLINE_SKILL_RE.search(body) else body
    # Prefer instruction + path; else any skill-path mention in body is enough
    if BODY_MUST_USE_RE.search(body_wo_inline):
        return True
    if SKILL_PATH_RE.search(body_wo_inline):
        return True
    return False


def needs_llm_skill_check(md: str) -> bool:
    """Ambiguous: mentions skill(s) but no clear frontmatter/path hit."""
    if has_frontmatter_skills(md) or has_body_skill_path(md):
        return False
    _, body = _split_md(md)
    if INLINE_SKILL_RE.search(body) and not SKILL_PATH_RE.search(body):
        # only inline → False, no LLM needed
        return False
    return bool(SKILL_WORD_RE.search(md))


class SkillPathUsage(BaseModel):
    uses_skill_files: bool = Field(
        description="True only if the agent must use skill FILE(S)/folder paths "
        "(e.g. .github/skills/.../SKILL.md). False for inline ## skill: only, "
        "or if skills are not required."
    )
    evidence: str = Field(default="", max_length=240)


SKILL_LLM_PROMPT = """You inspect a GitHub Agentic Workflow markdown file.

Does the BODY instruct the agent to use a skill as a FILE/FOLDER path
(e.g. `.github/skills/...`, `.agents/skills/.../SKILL.md`, `owner/repo/skills/...`)?

Answer uses_skill_files=true ONLY if the agent must read/follow/load such path skills.
Answer false for:
- inline skills only (`## skill: name`)
- mentioning the word "skill" without a path
- creating/writing a new SKILL.md as the *task output* (meta), unless it also loads an existing skill path

Return JSON: {"uses_skill_files": true/false, "evidence": "short quote"}
"""


def llm_body_uses_skill_path(text: str) -> SkillPathUsage:
    if not os.getenv("DEEPSEEK_API_KEY"):
        raise RuntimeError("Set DEEPSEEK_API_KEY in .env")
    # keep prompt small
    snippet = text[:5000]
    resp = completion(
        model=DEEPSEEK_MODEL,
        messages=[
            {"role": "system", "content": SKILL_LLM_PROMPT},
            {"role": "user", "content": snippet},
        ],
        response_format={"type": "json_object"},
        reasoning_effort="none",
        temperature=0,
        max_tokens=256,
        extra_body={"thinking": {"type": "disabled"}},
    )
    return SkillPathUsage.model_validate_json(resp.choices[0].message.content)


def _pair_key(repo: str, workflow: str) -> str:
    return f"{repo}::{workflow}"


def load_skills_done(path: Path) -> dict[str, dict]:
    done: dict[str, dict] = {}
    if not path.exists():
        return done
    with path.open(encoding="utf-8") as fh:
        for line in fh:
            line = line.strip()
            if not line:
                continue
            rec = json.loads(line)
            if rec.get("repo") and rec.get("workflow_name") and "uses_skills" in rec:
                if rec.get("error"):
                    continue
                done[_pair_key(rec["repo"], rec["workflow_name"])] = rec
    return done


print(f"model={DEEPSEEK_MODEL} concurrency={SKILLS_CONCURRENCY}")
print(f"resume={SKILLS_PROGRESS}")
```

```python
SKILLS_PROGRESS.parent.mkdir(parents=True, exist_ok=True)
done = load_skills_done(SKILLS_PROGRESS)

rows = []
llm_pending = []

for _, row in df_iss_triage.iterrows():
    pk = _pair_key(row["repo"], row["workflow_name"])
    if pk in done:
        rows.append(done[pk])
        continue

    md = str(row.get("md_content") or "")
    fm = has_frontmatter_skills(md)
    body = has_body_skill_path(md)
    if fm or body:
        rec = {
            "repo": row["repo"],
            "workflow_name": row["workflow_name"],
            "skills_frontmatter": fm,
            "skills_body_path": body,
            "uses_skills": True,
            "skills_detect": "frontmatter" if fm else "body_path_regex",
            "skills_evidence": "",
        }
        rows.append(rec)
    elif needs_llm_skill_check(md):
        llm_pending.append(row)
    else:
        rec = {
            "repo": row["repo"],
            "workflow_name": row["workflow_name"],
            "skills_frontmatter": False,
            "skills_body_path": False,
            "uses_skills": False,
            "skills_detect": "none",
            "skills_evidence": "",
        }
        rows.append(rec)

print(
    f"deterministic done={len(rows)} | llm pending={len(llm_pending)} | "
    f"resumed={len(done)}"
)

write_lock = threading.Lock()
new_n = 0
err_n = 0

def _classify_skills_row(row):
    md = str(row.get("md_content") or "")
    try:
        ans = llm_body_uses_skill_path(md)
        return {
            "repo": row["repo"],
            "workflow_name": row["workflow_name"],
            "skills_frontmatter": False,
            "skills_body_path": bool(ans.uses_skill_files),
            "uses_skills": bool(ans.uses_skill_files),
            "skills_detect": "body_path_llm",
            "skills_evidence": ans.evidence,
        }
    except Exception as exc:
        return {
            "repo": row["repo"],
            "workflow_name": row["workflow_name"],
            "skills_frontmatter": False,
            "skills_body_path": False,
            "uses_skills": False,
            "skills_detect": "error",
            "skills_evidence": "",
            "error": str(exc),
        }

with SKILLS_PROGRESS.open("a", encoding="utf-8") as prog:
    # persist deterministic rows not yet in done
    for rec in rows:
        pk = _pair_key(rec["repo"], rec["workflow_name"])
        if pk in done:
            continue
        prog.write(json.dumps(rec, ensure_ascii=False) + "\n")
        done[pk] = rec
        new_n += 1
    prog.flush()

    if llm_pending:
        with ThreadPoolExecutor(max_workers=SKILLS_CONCURRENCY) as pool:
            futs = [pool.submit(_classify_skills_row, r) for r in llm_pending]
            for fut in tqdm(as_completed(futs), total=len(futs), desc="skills llm"):
                rec = fut.result()
                pk = _pair_key(rec["repo"], rec["workflow_name"])
                with write_lock:
                    prog.write(json.dumps(rec, ensure_ascii=False) + "\n")
                    prog.flush()
                    if rec.get("error"):
                        err_n += 1
                    else:
                        done[pk] = rec
                        rows.append(rec)
                        new_n += 1

# Prefer latest successful records
records = list(load_skills_done(SKILLS_PROGRESS).values())
df_skills = pd.DataFrame(records)
df_skills.to_csv(SKILLS_OUT_CSV, index=False)

drop_cols = [
    c
    for c in (
        "skills_frontmatter",
        "skills_body_path",
        "uses_skills",
        "skills_detect",
        "skills_evidence",
    )
    if c in df_iss_triage.columns
]
df_iss_triage = df_iss_triage.drop(columns=drop_cols)
df_iss_triage = df_iss_triage.merge(
    df_skills[
        [
            "repo",
            "workflow_name",
            "skills_frontmatter",
            "skills_body_path",
            "uses_skills",
            "skills_detect",
            "skills_evidence",
        ]
    ],
    on=["repo", "workflow_name"],
    how="left",
)
df_iss_triage["uses_skills"] = df_iss_triage["uses_skills"].fillna(False).astype(bool)
df_iss_triage["skills_frontmatter"] = df_iss_triage["skills_frontmatter"].fillna(False).astype(bool)
df_iss_triage["skills_body_path"] = df_iss_triage["skills_body_path"].fillna(False).astype(bool)

print(f"new={new_n} errors={err_n} wrote {SKILLS_OUT_CSV}")
print("uses_skills:")
print(df_iss_triage["uses_skills"].value_counts(dropna=False))
print("\nby detect:")
print(df_iss_triage["skills_detect"].value_counts(dropna=False))
```
