```python
from pathlib import Path
from huggingface_hub import snapshot_download

# Repo layout is data/{artifacts,artifact_siblings,repos,mining_runs}/*.parquet
# Download into project root so paths become ./data/<config>/...
dest = snapshot_download(
    repo_id="mvaccargiu/gitskills",
    repo_type="dataset",
    local_dir=".",
    allow_patterns=["data/**"],
)
print("downloaded to", dest)

root = Path("data")
for p in sorted(root.iterdir()):
    if p.is_dir():
        n = len(list(p.glob("*.parquet")))
        size = sum(f.stat().st_size for f in p.rglob("*") if f.is_file())
        print(f"{p.name}: {n} parquet files, {size/1e9:.2f} GB")
```
